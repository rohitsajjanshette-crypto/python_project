```python
passenger = ["Rahul","Amit","Priya"]
seat = ["S1","S2","S3"]
fare = [2500, 1800, 3200]
```

```python
reservation_data = list(zip(passenger, seat, fare))
```

```python
print("---------------------------------------------")
print("-------Raiway Reservation System-------------")
print("---------------------------------------------")

for index, data in enumerate(reservation_data):
    passenger, seat, fare = data
    final_fare = (lambda x : x - 100)(fare)

    print("Reservation ID : ", index)
    print("Passenger      : ", passenger)
    print("Seat number    : ", seat)
    print("Final Fare     : ", final_fare)
    print("---------------------------------------------")
```
