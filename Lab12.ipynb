```python
def reservation_status(passenger_name, ticket_status = "Confirmed"):
    print("Passenger Name  : ", passenger_name)
    print("Ticket_status   : ", ticket_status)
```

```python
reservation_status("Rahul Sharma")
reservation_status("Amit Verma", "Waiting")
```

```python
def passenger_booking(*passengers):
    print("Passenger's List :")
    for name in passengers:
        print("Passenger Name : ", name)
```

```python
passenger_booking("Rahul Sharma","Amit Verma","Priya singh","Neha Kapoor")
```

```python
def ticket_details(**details):
    print("Ticket Information")
    for key, value in details.items():
        print(key, " : ", value)
```

```python
ticket_details(passenger_name = "Rahul Sharma", train_name = "Rajdhani Express", seat_number = "B1-21", ticket_status = "Confirmed")
```

```python
def seat_countdown(seat):
    if seat == 0:
        print("All seats are booked")
        return
    print("Remaining Seats : ", seat)
    seat_countdown(seat - 1)
```

```python
seat_countdown(5)
```

```python
def verifying_passenger(passenger_number):
    if passenger_number == 0:
        print("Verification completed")
        return
    print("Verifying Passenger :", passenger_number)
    verifying_passenger(passenger_number - 1)
```

```python
verifying_passenger(3)
```

```python
def final_booking(train_name = "Rajdhani Express", *seat_number):
    print("-------------------------------------")
    print("------Indian Railway Reservation-----")
    print("-------------------------------------")
    print("Train Name : ", train_name)
    for seat in seat_number:
        print("Allocated Seat Number : ", seat)
```

```python
final_booking("Rajdhani Express","B1-21","B1-22","B1-23")
```
