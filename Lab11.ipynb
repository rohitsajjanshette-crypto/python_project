```python
def calculate_fare(total_price, passengers):
    total_amount = total_price * passengers
    return total_amount
```

```python
final_fare = calculate_fare(2450, 2)
```

```python
print(final_fare)
```

```python
def verify_passenger(age):
    if age > 18:
        print("Passenger is eligible for Reservation")
    else:
        print("Not eligible")
```
